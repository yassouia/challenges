Let's look at the frequency of the characters in our dataset.

```python
# Preprocess/Cle
```

```python
# Get character frequencies

def get_char_counts(ds):
    all_text = ds['raw_text']
    char_counts = Counter(''.join(all_text))
    # Sort
    char_counts = dict(sorted(char_counts.items(), key=lambda item: item[1], reverse=True))
    return char_counts

char_counts = get_char_counts(ds)
char_counts
```

Now we can define the vocabulary of our TTS model.

```python
# Punctuation
PUNCT = set(' !،.:؟')

CHARS = char_counts.keys()
CHARS = set(CHARS)

# Remove punctuation from character set
CHARS = CHARS - PUNCT
CHARS = sorted(list(CHARS))
PUNCT = sorted(list(PUNCT))

print("Punctuation: ", PUNCT)
print("Character set: ", CHARS)

# Create character config
from TTS.tts.configs.shared_configs import CharactersConfig

characters = CharactersConfig(
    characters_class="TTS.tts.models.vits.VitsCharacters",
    pad="_",
    eos="&",
    bos="*",
    blank=None,
    characters="".join(CHARS),
    punctuations="".join(PUNCT),
)
```

# Prepare dataset
Let's store the dataset in a format that can be used with coqui.

We will create a csv file with with 4 columns:
- The relative path of the WAV audio file.
- The second column can be left empty.
- The transcription: the text spoken in the audio clip.
- The unique name of the speaker. This is useful for multi-speaker TTS.

```python
import pandas as pd
import soundfile as sf
import os
from tqdm.notebook import tqdm

# Create the output directory if it doesn't exist
data_output_dir = "ary_tts_dataset"
wavs_dir = os.path.join(data_output_dir, "wavs")
os.makedirs(wavs_dir, exist_ok=True)

# Create the metadata file
metadata_file = os.path.join(data_output_dir, "metadata.csv")
metadata = []

# Iterate over the dataset
for i, sample in tqdm(enumerate(ds), total=len(ds)):
    # Get the audio data, text, and speaker ID
    audio_array = sample['audio']['array']
    text = sample['raw_text']
    speaker = sample['speaker_id']

    # Create a unique file name
    file_name = f"{i:05d}.wav"
    file_path = os.path.join(wavs_dir, file_name)

    # Save the audio file
    sf.write(file_path, audio_array, SAMPLE_RATE)

    # Append to the metadata
    metadata.append(f"wavs/{file_name}||{text}|{speaker}|")

# Write the metadata to the csv file
with open(metadata_file, 'w') as f:
  for line in metadata:
    f.write(line + '\n')
```

```python
# Take a look at the metadata file
!head $data_output_dir/metadata.csv
```

```python
# Now we can define our dataset configuration so that coqui knows where the data and metadata are stored
dataset_conf = BaseDatasetConfig(
    formatter="brspeech", meta_file_train="metadata.csv", path=data_output_dir, dataset_name="ary_tts_dataset", language=ISO
)
```

# Calculate speaker embeddings
Embeddings are vector representations of an audio file. The TTS model takes this vector as input to try to mimic the original speaker's voice, this way we can clone any voice by first calculating the embedding then feeding it to the model.

We calculate embeddings for each sample in our dataset using a pretrained speaker encoder.

```python
### Extract speaker embeddings
SPEAKER_ENCODER_CHECKPOINT_PATH = (
    "https://github.com/coqui-ai/TTS/releases/download/speaker_encoder_model/model_se.pth.tar"
)
SPEAKER_ENCODER_CONFIG_PATH = "https://github.com/coqui-ai/TTS/releases/download/speaker_encoder_model/config_se.json"

D_VECTOR_FILES = []  # List of speaker embeddings/d-vectors to be used during the training

# Checks if the speakers embeddings are already computed, if not compute it
embeddings_file = os.path.join(OUT_PATH, "speakers.pth")
if not os.path.isfile(embeddings_file):
    print(f">>> Computing the speaker embeddings for the {dataset_conf.dataset_name} dataset")
    compute_embeddings(
        SPEAKER_ENCODER_CHECKPOINT_PATH,
        SPEAKER_ENCODER_CONFIG_PATH,
        embeddings_file,
        formatter_name=dataset_conf.formatter,
        dataset_name=dataset_conf.dataset_name,
        dataset_path=dataset_conf.path,
        meta_file_train=dataset_conf.meta_file_train,
        meta_file_val=dataset_conf.meta_file_val,
    )
D_VECTOR_FILES.append(embeddings_file)
```

# Define hyperparameters
Here we will define various hyperparameters of the VITS architecture the model is based on. Keep in mind that in order for the finetuned model to generate intelligible speech, it usually needs many epochs.

```python
# Audio config used in training.
audio_config = VitsAudioConfig(
    sample_rate=SAMPLE_RATE,
    hop_length=256,
    win_length=1024,
    fft_size=1024,
    mel_fmin=0.0,
    mel_fmax=None,
    num_mels=80,
)
```

```python
# Init VITSArgs setting the arguments that are needed for the YourTTS model
model_args = VitsArgs(
    spec_segment_size=62,
    hidden_channels=192,
    hidden_channels_ffn_text_encoder=768,
    num_heads_text_encoder=2,
    num_layers_text_encoder=10,
    kernel_size_text_encoder=3,
    dropout_p_text_encoder=0.1,
    d_vector_file=D_VECTOR_FILES,
    use_d_vector_file=True,
    d_vector_dim=512,
    speaker_encoder_model_path=SPEAKER_ENCODER_CHECKPOINT_PATH,
    speaker_encoder_config_path=SPEAKER_ENCODER_CONFIG_PATH,
    resblock_type_decoder="2",
    use_speaker_encoder_as_loss=False,
    # Useful parameters to enable multilingual training, in our case we only have one
    use_language_embedding=True,
    embedded_language_dim=4,
)
```

Here we define some sentences we evaluate the model on during training. Make sure you don't have OOV (out-of-vocabulary) characters!

You can listen to how the model synthesized them using Tensorboard, which we'll do at the end.

```python
TEST_SENTENCES = [
    ["هادي جملة بالدارجة بصوت مرا", "spk03", None, "ary"],
    ["هادي جملة بالدارجة بصوت راجل", "spk02", None, "ary"],
]
```

```python
# General training config, here you can change the batch size, the number of epochs and other useful parameters
config = VitsConfig(
    epochs=2, # The model needs more epochs to produce intelligible speech
    output_path=OUT_PATH,
    model_args=model_args,
    run_name=RUN_NAME,
    project_name="YourTTS",
    run_description=f"""
            - YourTTS trained on Darija speech.
        """,
    dashboard_logger="tensorboard",
    logger_uri=None,
    audio=audio_config,
    batch_size=BATCH_SIZE,
    batch_group_size=4,
    eval_batch_size=BATCH_SIZE,
    num_loader_workers=4,
    # eval_split_max_size=256,
    print_step=25,
    plot_step=50,
    # log_model_step=1000,
    save_step=1000,
    save_n_checkpoints=2,
    save_checkpoints=True,
    target_loss="loss_1",
    print_eval=True,
    compute_input_seq_cache=True,
    add_blank=True,
    text_cleaner="no_cleaners",
    characters=characters,
    phoneme_cache_path=None,
    precompute_num_workers=12,
    start_by_longest=True,
    datasets=[dataset_conf],
    cudnn_benchmark=False,
    min_audio_len=int(SAMPLE_RATE * MIN_AUDIO_LEN_IN_SECONDS),
    max_audio_len=SAMPLE_RATE * MAX_AUDIO_LEN_IN_SECONDS,
    mixed_precision=True,
    test_sentences=TEST_SENTENCES,
    speaker_encoder_loss_alpha=9.0,
)
```

```python
# Load all the datasets samples and split training and evaluation sets
train_samples, eval_samples = load_tts_samples(
    config.datasets,
    eval_split=True,
    eval_split_max_size=config.eval_split_max_size,
    eval_split_size=config.eval_split_size,
)
print(f"Loaded {len(train_samples)} train samples")
print(f"Loaded {len(eval_samples)} eval samples")
```

# Train the model

```python
# Init the model
model = Vits.init_from_config(config)

# Init the trainer and 🚀
trainer = Trainer(
    TrainerArgs(restore_path=RESTORE_PATH, skip_train_epoch=SKIP_TRAIN_EPOCH),
    config,
    output_path=OUT_PATH,
    model=model,
    train_samples=train_samples,
    eval_samples=eval_samples,
)
trainer.fit()
```
